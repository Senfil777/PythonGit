```python
res = input("введи через пробел суммы покупок:")
res_li = res.split(" ")

new_float_li = []

for elem in res_li:
    new_float_li.append(float(elem))

print(new_float_li)
print("summa:", sum(new_float_li))
```

```python
float_li = [float(elem) for elem in input("введи через пробелы суммы покупок:").split(" ")]

print(float_li)
print("summa:", sum(float_li))
```
