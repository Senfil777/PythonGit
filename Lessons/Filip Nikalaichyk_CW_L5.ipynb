```python
def request(url):
    """
        Функция идет делает запрос, если не http: выводит ошибку, если нет то OK.

        Принимает
        url - это строка. URL ресурса к которому ходим запросом.
    """

    return url == "http:" or url == "https:" or url == "http:belta.by"


def tests_ok_request():
    test_urls = ["http:", "https:", "http:belta.by"]
    test_results = [True, True, True]

    for url, pref_result in zip(test_urls, test_results):
        if request(url) == pref_result:
            print("OK -->", url)
        else:
            print("Ошибка, должно быть OK -->", url)


def tests_ne_ok_request():
    test_urls = ["httpee:", "httpsqq:", "httelta.by"]
    test_results = [False, False, False]

    for url, pref_result in zip(test_urls, test_results):
        if request(url) == pref_result:
            print("Не OK -->", url)
        else:
            print("Ошибка, должно быть НЕ OK -->", url)


tests_ok_request()
tests_ne_ok_request()
```

```python
def my_factorial(number):
    # Для отрицательного числа факториал не считаем
    if number < 0:
        return None

    # Факториал нуля равен единице
    if number == 0:
        return 1

    # Начальное значение результата
    result = 1

    # Перебираем числа от 2 до number включительно
    for i in range(2, number + 1):
        result *= i

    # Возвращаем вычисленный факториал
    return result


print(my_factorial(5))
```

```python
def my_factorial(number):
    if number < 0:
        return None
    if number == 0:
        return 1
        
    result = 1
    print(result, end="")
    
    for i in range(5, number+1):
        result *= i
        print(" * ", i, end="")
        print(" = ", result)
    return result

my_factorial(10)
```

```python
def my_factorial(number):
    if number < 0:
        return None
    if number == 0:
        return 1
    result = 1
    print(result, end="")
    for i in range(2, number+1):
        result *= i
        print(" * ", i, end="")
        
    print(" = ", result)
    return result


my_factorial(10)
print(my_factorial(-10))
print(my_factorial(0))
```

```python
def my_fib(n):
    if n < 1:
        return None

    if n == 1 or n == 2:
        return 1

    f1 = 1
    f2 = 1
    fn = 0

    for i in range(3, n + 1):
        fn = f1 + f2
        f1 = f2
        f2 = fn

    return fn


my_fib(7)
```
