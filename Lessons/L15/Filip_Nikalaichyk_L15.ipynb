```python
import sqlite3
import configparser as conf


cfg = conf.ConfigParser()
cfg.read("configs.cfg")


class UserModel:
    def __init__(self):
        self.c = sqlite3.connect(cfg["db"]["db_name"])
        self.cur = self.c.cursor()
        q_text = """CREATE TABLE IF NOT EXISTS "user" (
         "u_id" INTEGER NOT NULL UNIQUE,
         "u_name" TEXT NOT NULL,
         PRIMARY KEY("u_id" AUTOINCREMENT));""";
        self.cur.execute(q_text)
        self.c.commit()

    def read_all(self):
        self.cur.execute("select * from user")
        res = self.cur.fetchall()
        for user in res:
            print(user)

    def insert_user(self):
        u = input("Имя пользователя:")
        data = (u,)
        self.cur.execute("INSERT INTO user (u_name) VALUES (?);", data)
        self.c.commit()
    
    def delete_user(self):
        u = int(input("ID пользователя для удаления:"))
        data = (u,)
        self.cur.execute("DELETE FROM user WHERE u_id=(?);", data)
        self.c.commit()
```

```python
import peewee as p
```

```python
db = p.SqliteDatabase ("db_orm_example.db")
```

```python
class User(p.Model):
    user_name = p.CharField()
    user_age = p.IntegerField()

    class Meta:
        database = db
        db_table = "users"
```
