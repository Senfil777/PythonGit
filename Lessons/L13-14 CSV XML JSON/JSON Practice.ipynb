```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def import_user(cls, jstring):
        print("import to user from json")
        di = j.loads(jstring)
        return cls(di['name'], di['age'])

    def __str__(self):
        return f'str: {self.name}, {self.age}'
        
    def __repr__(self):
        return f'repr: {self.name}, {self.age}'

    def export_user(self):
        print("export user to json")
        return j.dumps(self.__dict__)

    
Dima = User("Dima", 123)
print(Dima)

jstr = Dima.export_user()
print(jstr, type(jstr))


New_dima = User.import_user(jstr)
print(New_dima, type(New_dima))
```
