```python
number_1 = float(input("Введите первое число: "))
number_2 = float(input("Введите второе число: "))
```

```python
#Выполняем операции
```

```python
if operation == "+":
    result = number_1 + number_2
    print("Результат:", result)
    
elif operation == "-":
    result = number_1 - number_2
    print("Результат:", result)

elif operation == "*":
    result = number_1 * number_2
    print("Результат:", result)

elif operation == "/":
    if number_2 == 0:
        print("Ошибка: на ноль делить нельзя")
    else:
        result = number_1 / number_2
        print("Результат:", result)

elif operation == "**":
    result = number_1 ** number_2
    print("Результат:", result)

else:
    print("Такой операции нет")
```
