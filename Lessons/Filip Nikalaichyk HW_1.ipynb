```python
# Вводим четыре числа
a = int(input("Введите первое число: "))
b = int(input("Введите второе число: "))
c = int(input("Введите третье число: "))
d = int(input("Введите четвёртое число: "))

# Определяем самое большое число
if a >= b:
    largest1 = a
else:
    largest1 = c

if c >= d:
    largest2 = c
else:
    largest2 = d

if largest1 >= largest2:
    largest = largest1
else:
    largest = largest2
print('Самое большое число: ', largest)
```
