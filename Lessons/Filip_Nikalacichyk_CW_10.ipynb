```python
class User:
    def __init__(self, name, phone):
        self.__name = name
        self.phone = phone

    def get_name(self):
        return self.__name

    def set_name(self, uname):
        if isinstance(uname, str):
            self.__name = uname
            return True

        return "Не верное имя пользователя"

    def __repr__(self):
        return f"repr\n\tname: {self.__name}\n\tphone: {self.phone}"

    def __str__(self):
        return f"\n\tname: {self.__name}\n\tphone: {self.phone}"


dima = User("Dima", 1212421412)

print(dima.set_name("Dimas"))
print(dima.get_name())
print(dima)
print()
dima
```

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __len__(self):
        return self.age

    def __add__(self, diff_car):
        return self.age + diff_car.age


vasia = Cat("Vasia", 15)

print(len(vasia))

vasia = Cat("Vasia", 15)
dima = Cat("Dima", 45)

print(vasia + dima)
```

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __len__(self):
        return self.age

    def __add__(self, diff_cat):
        return self.age + diff_cat.age

    def __pow__(self, power):
        # поддержим возведение в степень только для чисел
        return self.age ** power


vasia = Cat("Vasia", 15)
print(len(vasia))

vasia = Cat("Vasia", 15)
dima = Cat("Dima", 45)

print(vasia + dima)

print(vasia ** 3)
```
