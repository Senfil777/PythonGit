```python
tasks = []

menu = """
________________________________________________
To Do List v0.20

Меню:
1 - добавить задачу
2 - показать все задачи
3 - отметить задачу выполненной
4 - удалить задачу
0 - выйти из программы
________________________________________________
"""


def add_task():
    task_name = input("Введи новую задачу: ")

    task = {
        "name": task_name,
        "done": False
    }

    tasks.append(task)

    print("Задача добавлена!")


def show_tasks():
    if len(tasks) == 0:
        print("Список задач пуст!")
        return

    print("Список задач:")

    for index, task in enumerate(tasks, start=1):

        if task["done"] == True:
            status = "Выполнено"
        else:
            status = "Не выполнено"

        print(index, "-", task["name"], "-", status)


def complete_task():
    if len(tasks) == 0:
        print("Список задач пуст!")
        return

    show_tasks()

    task_number = input("Введи номер выполненной задачи: ")

    if task_number.isdigit():
        task_number = int(task_number)

        if task_number >= 1 and task_number <= len(tasks):
            tasks[task_number - 1]["done"] = True
            print("Задача отмечена выполненной!")
        else:
            print("Задачи с таким номером нет!")

    else:
        print("Нужно ввести номер задачи!")


def delete_task():
    if len(tasks) == 0:
        print("Список задач пуст!")
        return

    show_tasks()

    task_number = input("Введи номер задачи для удаления: ")

    if task_number.isdigit():
        task_number = int(task_number)

        if task_number >= 1 and task_number <= len(tasks):
            deleted_task = tasks.pop(task_number - 1)
            print("Удалена задача:", deleted_task["name"])
        else:
            print("Задачи с таким номером нет!")

    else:
        print("Нужно ввести номер задачи!")


def run_app():
    print(menu)

    operation_number = input("---> ")

    while operation_number != "0":

        if operation_number == "1":
            add_task()

        elif operation_number == "2":
            show_tasks()

        elif operation_number == "3":
            complete_task()

        elif operation_number == "4":
            delete_task()

        else:
            print("Не понял операцию. Введи номер из меню.")

        print()
        print(menu)

        operation_number = input("---> ")

    print("Программа завершила работу! Пока.")


run_app()
```
