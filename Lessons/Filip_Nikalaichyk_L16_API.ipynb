```python
import requests as r


api_link = "https://api.nbrb.by/exrates/rates/"
currencies = {"usd":"431", "NZD":"448", "KRW":"470"}

balance = 1000
nzdollar_str = f'{api_link}{currencies.get("NZD")}'


nzdollar_api_result = r.get(nzdollar_str)

print(nzdollar.text, end="\n\n")
print(nzdollar.url, end="\n\n")
print(nzdollar.headers, end="\n\n")
print(nzdollar.status_code, end="\n\n")

dict_nzd = nzdollar.json()
print(dict_nzd, end="\n\n")

print("BYN:", balance)
print("NZD:", balance/dict_nzd.get('Cur_OfficialRate'))
```

```python
# API github

import requests as r


api_link = "https://api.github.com/users/"
user_name = "pallets"

url = f'{api_link}{user_name}'


api_result = r.get(url)

print(api_result.text, end="\n\n")
print(api_result.url, end="\n\n")
print(api_result.headers, end="\n\n")
print(api_result.status_code, end="\n\n")

dict_user_github = api_result.json()
print(dict_user_github, end="\n\n")

print("login:", dict_user_github.get('login'))
print("blog:", dict_user_github.get('blog'))
print("email:", dict_user_github.get('email'))
print("url:", dict_user_github.get('url'))
print("avatar_url:", dict_user_github.get('avatar_url'))
```

```python
# API NBRB

import requests as r


api_link = "https://api.nbrb.by/exrates/rates/"
currencies = {"USD":"431", "NZD":"448", "KRW":"470"}

balance = 1000


for currency_name, currency_code in currencies.items():
    url = f'{api_link}{currency_code}'
    api_result = r.get(url)

    print(api_result.url)
    print(api_result.status_code)

    currency_res = api_result.json()

    print("BYN:", balance)
    print(currency_name, balance / currency_res.get('Cur_OfficialRate'))
    print()
```
