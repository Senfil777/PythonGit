```python
Предустановленные модули

Модуль — это отдельный файл на Python, в котором находятся различные функции, типы данных и так далее.

Чтобы всем этим пользоваться, вам нужно подключить этот файл к своей программе.

У модуля есть имя.
Сначала необходимо узнать имя нужного модуля.
Далее мы подключаем его к нашей программе.
Затем ищем в нём нужные функции для наших задач.
Используем их.
```

```python
from datetime import date


date3 = date(1995, 2, 15)
date4 = date(2026, 7, 31)

print((date4 - date3).days)
```

```python
import datetime as dt


date7 = dt.date(1995, 2, 15)
date8 = dt.date(2026, 7, 31)

print((date8 - date7).days)
```

```python
from datetime import date as d

date10 = d(1995, 2, 15)
date20 = d(2026, 7, 31)

print((date20 - date10).days)
```
