```python
def is_year_leap(year):
    # Если год делится на 400 без остатка,
    # то год високосный
    if year % 400 == 0:
        return True

    # Если год делится на 100 без остатка,
    # то год не високосный
    if year % 100 == 0:
        return False

    # Если год делится на 4 без остатка,
    # то год високосный
    if year % 4 == 0:
        return True

    # Во всех остальных случаях год не високосный
    return False


# Готовые данные для проверки функции
test_data = [1500, 1900, 2000, 2016, 1987]
test_results = [False, False, True, True, False]


# Готовый код проверки
for year, result in zip(test_data, test_results):
    if is_year_leap(year) == result:
        print(year, "is leap? -->", result)
    else:
        print(year, "from your function -->",
              is_year_leap(year))
        print("but expected -->", result)
```
