```python
class User:
    def __init__(self, name):
        # self ==== переменная экземпляра
        self._name = name

    @property
    def name(self):
        print("get name")
        return self._name
    
    @name.setter
    def name(self, uname):
        print("set name")
        if isinstance(uname, str):
            self._name = uname
        print("new name is - ", self._name)

u12 = User("Dima")
print(u12.name)
u12.name = "sdasda"
```

```python
def pow_number_two(*args):
    for i in args:
        print(i ** 2)

pow_number_two(4, 2, 3, 4, 5)
```

```python
def dict_usage(**kwargs):
    print(type(kwargs))
    print(len(kwargs))
    print(kwargs)

    for k, v in kwargs.items():
        print("key:", k, "   - value:", v)


dict_usage(name="Stef", age=12323, phone=1323123123)
```

```python
class User:
    def __init__(self, name):
        self.name = name

    def __iter__(self):
        return iter(self.name)

u12 = User("Dima")
for i in u12:
    print(i)
```

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades
    def __iter__(self):
        print("Оценки студента - ", self.name)
        return iter(self.grades)

Dima = Student("Dima", [6, 7, 9, 4, 10])
for i in Dima:
    print(i)
```

```python
def n(x):
    for i in range(x):
        yield i**2/2*4

for i in n(15):
    print(i)
```
