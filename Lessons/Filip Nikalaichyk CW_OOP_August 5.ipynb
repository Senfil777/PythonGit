```python
class Wallet:
    def __init__(self, wid, currency, volume):
        self.wid = wid
        self.currency = currency
        self.volume = volume

    def __str__(self):
        return f"wid:{self.wid}, currency: {self.currency}, volume:{self.volume}"

    def __repr__(self):
        return f"repr_wid:{self.wid}, repr_currency: {self.currency}, volume:{self.volume}"
        
```

```python
class Offline_Wallet:
    def __init__(self, wid, currency, volume):
        self.wid = wid
        self.currency = currency
        self.volume = volume

        
    def set_currency (self, new_currency): 
        self.currency = new_currency
        print ("Валюта измненена")

    
    def __str__(self):
        return f"wid:{self.wid}, currency: {self.currency}, volume:{self.volume}"

    
    def __repr__(self):
        return f"repr_wid:{self.wid}, repr_currency: {self.currency}, volume:{self.volume}"


F_Wallet = Wallet (123445, "BYN", 1234567890)
        
```

```python
class Online_Wallet1(Offline_Wallet):
    def __init__(self, wid, currency, volume, secret):
        self.secret = secret
        super().__init__(wid, currency, volume)
        
    def __str__(self):
        return f"wid:{self.wid}, currency: {self.currency}, volume:{self.volume}, secret: {self.secret}"

    
    def __repr__(self):
        return f"repr_wid:{self.wid}, repr_currency: {self.currency}, volume:{self.volume}, secret: {self.secret}"
    pass

MyWallet = Online_Wallet1 (1, "USD", 1000, 500)
```
