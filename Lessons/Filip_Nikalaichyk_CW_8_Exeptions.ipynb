```python
try:
    #li = [1, 2, 3]
    #li.epend(312)
    #1 / 0

    #hello + 9
    #prin()
    #124124 / "sdfsd"
    4 + 6

except ZeroDivisionError:
    print("атата не дели на ноль!")

except AttributeError:
    print("Будь внимателен с именами методов!")

except:
    print("упс, что-то пошло не так")

else:
    print("ошибок не было")
```

```python
finally - отрабатывает всегда
else - отрабатывает если не было ошибок
except - отрабатывает для ошибок чтобы их перехватить и обработать
try - отрабатывает всегда, в ней какойто рискованный код
```

```python
try:
    name = input()

except ZeroDivisionError:
    print("атата не дели на ноль!")

except AttributeError:
    print("Будь внимателен с именами методов!")

except:
    print("упс, что-то пошло не так")

else:
    print("ошибок не было")

finally:
    print("я работаю всегда")
```
