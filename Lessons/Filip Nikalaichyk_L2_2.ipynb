```python
pin = 1234

match pin:
case 1234:
    print("Привет, вход выполнен admin!")
case 123:
    print('Привет, вход выполнен manager!')
case 12:
    print('Привет, вхо"д выполнен u"ser!')
case _:
    print('Вход не выполнен. попробуй ещё раз!')
```

```python
numbers = int(input("сколько он хочет ввести чисел"))
summa = 0
counter = numbers

while counter > 0:
    summa += int(input("Введи целое число:"))
    counter -= 1


print("Сумма и количество:")
print(summa)
print(numbers)
print("Считаю среднее:")
print(summa/numbers)
```
