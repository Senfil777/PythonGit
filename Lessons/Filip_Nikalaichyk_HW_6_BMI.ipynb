```python
menu = """
________________________________________

BMI Calculator

Введите:
- вес в килограммах
- рост в метрах

Для выхода - 0.
________________________________________
"""


def calculate_bmi(weight, height):
    """
    Функция рассчитывает индекс массы тела

    Если вес или рост указаны неправильно,
    функция возвращает None.
    """

    if height < 1.0 or height > 2.5:
        return None

    if weight < 20 or weight > 200:
        return None

    bmi_result = weight / height ** 2

    return bmi_result


def get_bmi_status(bmi_result):
    """
    Функция получает индекс массы тела
    и возвращает текстовую категорию.
    """

    if bmi_result < 18.5:
        return "Дефицит массы тела"

    elif bmi_result < 25:
        return "Норма"

    elif bmi_result < 30:
        return "Избыточная масса тела"

    else:
        return "Ожирение"


def run_app():
    print(menu)

    weight = float(input("Введите вес в килограммах: "))
    height = float(input("Введите рост в метрах: "))

    while weight != 0 and height != 0:

        result = calculate_bmi(weight, height)

        if result is None:
            print("Вес или рост введены неправильно.")

        else:
            result = round(result, 2)
            status = get_bmi_status(result)

            print()
            print(f"Ваш BMI: {result}")
            print(f"Результат: {status}")

        print()
        print("Для выхода введите 0.")

        weight = float(input("Введите вес в килограммах: "))
        height = float(input("Введите рост в метрах: "))

    print("Программа завершила работу!")


run_app()
```
