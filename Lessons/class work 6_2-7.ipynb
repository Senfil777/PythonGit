```python
Откуда питон(пайтон) берет функции:
    
1 встроенные (Input, int, print)

2 предустановленные модули (import имя_модуля)

3 те что пишем мы (def ....)


4 докачать с хранилки не типовые модули (для тестирования, дата сайнса, бэкенда)
```

```python
предустановленные модули
модуль это - отдельный файл на пайтон в 
котором лежат какието не типовые функции, 
типы данных и тп

чтобы всем этим пользоваться - вам нужно
этот файл подключить к своей программе

1 что у модуля есть ИМЯ
2 перед всем должны узнать имя модуля который нам нужен
3 далее мы его подключаем к нашей программе
4 далее ищем в нем нужные функции для наших задач
5 и используем их
```

```python
import datetime


date1 = datetime.date(1995, 2, 15)
date2 = datetime.date(2026, 7, 31)

print((date2 - date1).days)
```

```python
from datetime import date


date1 = date(1995, 2, 15)
date2 = date(2026, 7, 31)

print((date2 - date1).days)
```

```python
from datetime import date as d


date1 = d(1995, 2, 15)
date2 = d(2026, 7, 31)

print((date2 - date1).days)
```

```python
import datetime as dt


date1 = dt.date(1995, 2, 15)
date2 = dt.date(2026, 7, 31)

print((date2 - date1).days)
```
