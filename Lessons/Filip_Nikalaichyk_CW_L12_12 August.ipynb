```python
class A:
    def __init__(self):
        self.a = 100
        
    def fun_a(self):
        return "A"


class B(A):
    def __init__(self):
        super().__init__()
        self.b = 200
        
    def fun_b(self):
        return "B"


bb = B()
print(bb.b)
print(bb.fun_b())
print(bb.fun_a())
print(bb.a)
```

```python
# MRO
порядок разрешения/расширения методв
ООП!

При наследовании важно это понимать
При еденичном наследовании 
class User(People)
- ищет вначале у самого себя (User + self), 
потом идет к родителю (People), потом к его родителю  и так далее 
если не нашел метод или свойство - ошибка - AttributeError

При множественном наследовании
class User(People, Animal)
ищет вначале у себя (User + self)
потом у первого родителя (People), потом у второго (Animal)
если не нашел метод или свойство - ошибка - AttributeError
```

```python
class UserNameError(Exception):
    """
    Ты не можешь вводить имя которе не строка.
    """

    pass


class User:
    counter = 0  
    def __init__(self, name, phone):
        # self ==== переменная экземпляра
        self.__name = name
        self.phone = phone
        User.counter += 1

    def get_name(self):
        return self.__name
    
    def set_name(self, uname):
        if isinstance(uname, str):
            self.__name = uname
            return True
            
        raise UserNameError("Имя пользователя введено не верно!")
  
    def __repr__(self):
        return f"repr\n\tname:{self.__name}\n\tphone:{self.phone}"

    def __str__(self):
        return f"\n\tname:{self.__name}\n\tphone:{self.phone}"


user1 = User("Kat", 2141212412)
print(user1)
user1.set_name("Dima")


print(user1)
try:
    user1.set_name(123123)
except UserNameError as e:
    print(e.args)

print(user1)

```
