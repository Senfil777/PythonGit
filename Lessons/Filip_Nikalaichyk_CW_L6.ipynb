```python
#имя - его оценки

#имя - str
#его оценки - int -


#оценок много, чтобы все их хранить для 1 студента нужен list/tuple

#имя оценок много
#str:list/tuple


#Словарь:

#{
#"ИМЯ1": [1,2,3,4,5],
#"ИМЯ2": [1,2,3,4,5],
#"ИМЯ3": [1,2,3,4,5],
#}

#Меню:
#1 - добавить нового студента - add_student_by_name - внесем имя и пустой список
#2 - добавить оценку студенту - add_grade_by_name
#3 - считаем средний балл - calc_avg_by_name
#0 - выйти из программы
```

```python
school = {}

menu = """

School v1

Меню:
1 - добавить нового студента - add_student_by_name - внесем имя и пустой список
2 - добавить оценку студенту - add_grade_by_name
3 - считаем средний балл - calc_avg_by_name
0 - выйти из программы
"""

print(menu)


```

```python
operation_number = input ("---->")

while operation_number != "0":
    if operation_number == "1":
        pass
    elif operation_number == "2":
        pass
    elif operation_number == "3":
        pass   
    elif operation_number == "4":
        pass
    else:
        print ("Не понял, повторяй")
        print (menu)
    operation_number = input ("---->")
```

```python
def add_grade_by_name():
    name = input("Введи имя студента: ")
    grade = int(input("Введи оценку студента: "))

    if name in school:
        school[name].append(grade)
    else:
        school[name] = [grade]


def show_school():
    for name, grades in school.items():
        print(name, grades)


print(menu)
operation_number = input("---> ")

while operation_number != "0":

    if operation_number == "1":
        add_grade_by_name()

    elif operation_number == "2":
        pass

    elif operation_number == "3":
        show_school()

    else:
        print("Не понял операцию. Введи номер в соответствии с меню.")

    print()
    print(menu)
operation_number = input("---> ")


print("Программа завершила работу! Пока.")
```
